## 課題
### 単語の文字数をカウントしよう
#### 仕様
- 英単語の入力を受け付けます。
- 未入力でリターン（エンター）キーを押されるまで、英単語の受付を繰り返します。
- 受付が完了したら、入力された英単語をABC順に並べ替えて表示します。
- また、すべての英単語に含まれるアルファベットごとの数も表示します。

```python
# 英単語を格納する空のリスト
words = []

# アルファベットごとの数を格納する辞書を作成
alphabet_count = {
    'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0,
    'j':0, 'k':0, 'l':0, 'm':0, 'n':0, 'o':0, 'p':0, 'q':0, 'r':0,
    's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0
}

# 英単語の入力受付（未入力で終了）
while True:
    word = input("英単語を入力してください：")
    
    if word == "":
        break
    
    words.append(word)
    
    # 1文字ずつ確認してカウント（小文字に変換）
    for char in word.lower():
        if char in alphabet_count:
            alphabet_count[char] += 1

# 英単語をABC順に並べ替え
words.sort()

# 英単語の一覧表示
print(f"入力した英単語：{words}")

# アルファベットごとの出現回数を表示（0より大きいもののみ）
for key, value in alphabet_count.items():
    if value > 0:
        print(f"{key}が{value}個ありました")
```
