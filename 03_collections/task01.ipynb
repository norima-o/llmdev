## 課題
### 学生の成績分析ツールを作成しよう

#### 仕様
- 学生の成績をリストとして定義します。以下のソースコードを使用してください。
  - `grades = [85, 90, 78, 92, 88]`
- 成績のリストを使用して以下を行ってください。
  - 成績の平均点を計算します。
  - 成績の最高点と最低点を求めます。
  - 成績が平均以上の学生数をカウントします。
- 結果を出力してください。

```python
# 成績リスト
grades = [85, 90, 78, 92, 88]
# grades = [70, 75, 80, 85, 90]

# 平均点の計算
average = sum(grades) / len(grades)

# 最高点・最低点
highest = max(grades)
lowest = min(grades)

# 平均以上の学生数をカウント
count_above_average = 0
for grade in grades:
    if grade >= average:
        count_above_average += 1

# 結果の出力
print(f"平均点: {average:.1f}")
print(f"最高点: {highest}")
print(f"最低点: {lowest}")
print(f"平均以上の学生数: {count_above_average}")
```
