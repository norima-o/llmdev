## 料込価格を計算しよう

```python
# 税抜価格の入力
price = int(input("税抜価格を入力してください："))

# 消費税率
tax_rate = 0.10

# 税込価格の計算
tax_included_price = int(price * (1 + tax_rate))

# 送料の判定（税込価格が2000円以上で送料無料）
if tax_included_price >= 2000:
    shipping = 0
else:
    shipping = 350

# 送料込税込価格の計算
total_price = tax_included_price + shipping

# 結果の表示（f文字列を使用）
if shipping == 0:
    print("送料は無料です")
else:
    print(f"送料は{shipping}円です")

print(f"送料込税込価格は{total_price}円です")
```
