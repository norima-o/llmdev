### 10年後の金額を複利で予測するプログラムを作成しよう
- コンピュータは預金や運用の計算にも広く用いられています。ここでは、10年後の金額を複利で予測するツールを作成してみましょう。



```python
# 元金の入力
principal = float(input("元金を入力してください: "))

# 年利（％）の入力
rate = float(input("年利を入力してください: "))

# 複利計算（10年間）
amount = principal
for i in range(10):
    amount *= (1 + rate / 100)

# 結果の表示（単位付き）
print(f"元金: {principal:.2f}円")
print(f"年利: {rate:.2f}%")
print(f"10年後の金額: {amount:.2f}円")
```
