## RAG連動型チャットボットを作成しよう

```python
# 必要なモジュールをインポート
import os
from dotenv import load_dotenv

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.llms.openai import OpenAI
from llama_index.core.chat_engine import CondensePlusContextChatEngine

# 環境変数の取得
load_dotenv("../.env")
os.environ["OPENAI_API_KEY"] = os.environ["API_KEY"]

# モデル名
MODEL_NAME = "gpt-4o-mini"

# Indexの構築
documents = SimpleDirectoryReader("./data/text").load_data()
index = VectorStoreIndex.from_documents(documents)

# Retriever（top_k を調整したい場合）
retriever = index.as_retriever(similarity_top_k=3)

# LLM
llm = OpenAI(model=MODEL_NAME)

# Chat Engine（Condense + Context）
chat_engine = CondensePlusContextChatEngine.from_defaults(
    retriever=retriever,
    llm=llm,
    verbose=True,
)

# チャットの開始
while True:
    message = input("メッセージを入力: ")
    if message.strip() == "":
        break

    # 質問を表示
    print(f"\n質問: {message}")

    # 質問（= チャットエンジンに投げる）
    response = chat_engine.chat(message)

    # 回答を表示
    # response は ChatResponse 系で、本文は response.response に入ることが多いです
    answer_text = getattr(response, "response", None) or str(response)
    print(f"\n回答: {answer_text}\n")

print("\n---ご利用ありがとうございました！---")
```
